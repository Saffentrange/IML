# Modeling and Prediction
TODO: Define the model and fit it using training data. Then, use test data to make predictions

```python
"""
This defines the model, fits training data and then does the prediction
with the test data 

Parameters
----------
X_train: matrix of floats, training input with 10 features
y_train: array of floats, training output
X_test: matrix of floats: dim = (100, ?), test input with 10 features

Compute
----------
y_test: array of floats: dim = (100,), predictions on test set
"""
class Model(object):
    def __init__(self):
        super().__init__()
        self._x_train = None
        self._y_train = None
        
        self.scaler = StandardScaler()  # teh scaler we use
        self.model = None   # save the best model we found here

    def fit(self, X_train: np.ndarray, y_train: np.ndarray):
        #TODO: Define the model and fit it using (X_train, y_train)
        # scale the data
        X_scaled = self.scaler.fit_transform(X_train)

        kernels = {
            'DotProduct': C(1.0) * DotProduct(sigma_0=1.0) + WhiteKernel(noise_level=1e-2),
            'DotProduct + RBF': C(1.0) * DotProduct(sigma_0=1.0) + C(1.0) * RBF(length_scale=1.0) + WhiteKernel(noise_level=1e-2),
            'DotProduct + Matern': C(1.0) * DotProduct(sigma_0=1.0) + C(1.0) * Matern(length_scale=1.0, nu=1.5) + WhiteKernel(noise_level=1e-2)
        }

        best_score = -np.inf
        best_kernel = None

        # Cross-validate all the kernel-candidates form above to find the best fitting kernel
        for name, kernel in kernels.items():
            gpr = GaussianProcessRegressor(
                kernel=kernel,
                n_restarts_optimizer=10,
                normalize_y=True,
                alpha=1e-6,
                random_state=42
            )
            scores = cross_val_score(gpr, X_scaled, y_train, cv=5, scoring='r2', n_jobs=-1)
            mean_r2 = scores.mean()
            print(name, mean_r2)
            if mean_r2 > best_score:
                best_score = mean_r2
                best_kernel = name
        
        self.model = GaussianProcessRegressor(
            kernel=kernels[best_kernel],
            n_restarts_optimizer=10,
            normalize_y=True,
            alpha=1e-6,
            random_state=42
        )
        self.model.fit(X_scaled, y_train)

    def predict(self, X_test: np.ndarray) -> np.ndarray:
        #TODO: Use the model to make predictions y_pred using test data X_test
        # y_pred=np.zeros(X_test.shape[0])
        X_scaled = self.scaler.transform(X_test)
        y_pred = self.model.predict(X_scaled)
        assert y_pred.shape == (X_test.shape[0],), "Invalid data shape"
        return y_pred
```

```python
model = Model()
# Use this function to fit the model
model.fit(X_train=X_train, y_train=y_train)
# Use this function for inference
y_pred = model.predict(X_test)
print(y_pred)
```
